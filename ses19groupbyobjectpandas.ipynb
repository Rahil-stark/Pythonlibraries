```python
# aggregate method
# passing dict

genres = movies.groupby('Genre')
#mean means avg
genres.agg(
    {
        'Runtime' : 'mean',
        'IMDB_Rating' : 'mean',
        'No_of_Votes' : 'sum',
        'Gross' : 'sum',
        'Metascore' : 'min'
    }
)
```

```python
# genres.agg(['min', 'max', 'mean'],numeric_only=True)

movies.groupby('Genre').agg({
    'Runtime': ['min', 'max', 'mean'],
    'IMDB_Rating': ['min', 'max', 'mean'],
    'Gross': ['min', 'max', 'mean']
    
})
```

```python
# Looping on groups
genres = movies.groupby('Genre')
genres
#Find the highest rated movie of each genre
# df = pd.DataFrame(columns=movies.columns)
# print(df)
for group, data in genres:
    print(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])
```

```python
# split (apply) combine
#apply -> builtin function
genres = movies.groupby('Genre')
# genres.apply(min)
genres.min() # aise bhi likh sakte hai wo work nhi kar raha tha
```

```python
# Find number of movies starting with A for each group

# def later(group):
#     print(type(group))
#     return(group)
def later(group):
    return group['Series_Title'].str.startswith('A').sum()
genres.apply(later)

# genres.apply(later)
```

```python
# find ranking of each movie in the group according to IMDB score

def rank_movie(group):
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending=True)
    return group
```

```python
genres.apply(rank_movie)
```

```python
# Find normalized IMDB rating group wise

def normal(group):
    group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()/(group['IMDB_Rating'] - group['IMDB_Rating'].min()))
    return group
genres.apply(normal)
```
